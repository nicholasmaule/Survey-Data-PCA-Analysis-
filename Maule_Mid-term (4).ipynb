```python
maxAbsValue = np.max(np.abs(x))
```

```python
scaleFactor = 10 ** np.ceil(np.log10(maxAbsValue))
```

```python
decX = x/scaleFactor
decX
```

```python
from sklearn.decomposition import PCA
```

```python
df = pd.read_csv('SurveyData.csv')
x = df.loc[:,:].values
```

```python
pca = PCA(n_components = 12)
```

```python
principleComponents = pca.fit_transform(x)
```

```python
principleDf = pd.DataFrame(data = principleComponents, columns = ['PC1','PC2','PC3','PC4','PC5','PC6','PC7','PC8','PC9','PC10','PC11','PC12'])
```

```python
principleDf
```

```python
explained_variance = pca.explained_variance_ratio_

variance_df = pd.DataFrame({'Component': [f'PC{i+1}' for i in range(12)],'Explained Variance': explained_variance})

variance_df
```
